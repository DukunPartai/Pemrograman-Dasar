```python
#ALFREDO ZIZANI 4232511061 13/11/2026 PEMROGRAMAN DASAR
#SOAL NOMOR 2

# Menyimpan nilai ke dalam variabel
daya_kw = 450
jam = 8
energi_aktual = 7200
kapasitas_kw = 400

# Menghitung energi harian
energi = daya_kw * jam

# Menghitung capacity factor
capacity_factor = (energi_aktual / (kapasitas_kw * 24)) * 100

# Menampilkan hasil
print(f"Energi harian = {energi} kWh")
print(f"Capacity factor = {capacity_factor:.2f}%")

#Program menyimpan nilai daya, waktu operasi, energi aktual, dan kapasitas PLTD ke dalam variabel.
#Program menghitung energi harian dengan mengalikan daya dengan waktu operasi
#Program menghitung capacity factor dengan membagi energi aktual dengan kapasitas maksimal selama 24 jam, kemudian dikalikan 100%.
#Program menampilkan hasil menggunakan print() dan f-string.
```

```python
#ALFREDO ZIZANI 4232511061 13/11/2026 PEMROGRAMAN DASAR
#SOAL NOMOR 6

# Membaca input suhu dan tekanan
suhu = float(input("Masukkan suhu uap (°C): "))
tekanan = float(input("Masukkan tekanan uap (bar): "))

# Menentukan status boiler
if suhu >= 550 or tekanan >= 14:
    print("STATUS: TRIP")
elif suhu > 520 and tekanan > 12:
    print("STATUS: WARNING")
else:
    print("STATUS: NORMAL")

#input() digunakan untuk menerima nilai suhu dan tekanan dari pengguna.
#float() mengubah input menjadi tipe data angka desimal.
#Program memeriksa kondisi pertama. Jika suhu mencapai 550°C atau tekanan mencapai 14 bar, status menjadi TRIP.
#Jika kondisi pertama tidak terpenuhi, program memeriksa kondisi kedua. Jika suhu lebih dari 520°C dan tekanan lebih dari 12 bar, status menjadi WARNING.
#Jika kedua kondisi tidak terpenuhi, status menjadi NORMAL.
```

```python
#ALFREDO ZIZANI 4232511061 13/11/2026 PEMROGRAMAN DASAR
#SOAL NOMOR 8

# Nilai awal volume tangki
volume = 100

# Perulangan selama volume masih lebih dari 0
while volume > 0:
    volume -= 20

    # Menampilkan volume setelah dikurangi
    print(f"Volume: {volume} liter")

    # Memeriksa apakah volume mencapai 20 liter
    if volume == 20:
        print("WARNING: Hampir kosong!")

# Pesan setelah perulangan selesai
print("TANGKI KOSONG - Pengosongan berhenti.")

#Variabel volume diisi dengan nilai awal 100 liter.
#Perintah while volume > 0 menjalankan perulangan selama volume masih lebih dari 0.
#volume -= 20 mengurangi volume sebanyak 20 liter setiap perulangan.
#print() menampilkan volume setelah dikurangi.
#if volume == 20 memeriksa apakah volume tepat 20 liter. Jika benar, program menampilkan peringatan.
#Ketika volume mencapai 0, kondisi while menjadi salah sehingga perulangan berhenti.
#Program menampilkan pesan bahwa tangki sudah kosong.
```

```python
#ALFREDO ZIZANI 4232511061 13/11/2026 PEMROGRAMAN DASAR
#SOAL NOMOR 10

# Data suhu bearing
data = ["62.5", "58.0", "81.3", "N/A", "66.0"]

# Menghitung jumlah data valid
jumlah_valid = 0

# Memeriksa setiap data
for i in range(len(data)):
    try:
        suhu = float(data[i])
        jumlah_valid += 1

        # Memeriksa suhu bearing
        if suhu > 70.0:
            print(f"Warning: Titik {i+1} suhu bearing tinggi!")

    except ValueError:
        print(f"Titik {i+1}: data tidak valid - pemeriksaan dihentikan")
        break

# Menampilkan jumlah data valid
print(f"Jumlah data valid: {jumlah_valid}")

#Program menyimpan lima data suhu bearing dalam list data.
#Variabel jumlah_valid dimulai dari 0 untuk menghitung data yang berhasil diperiksa.
#Perulangan for memeriksa setiap data berdasarkan indeks.
#float() mengubah data teks menjadi angka desimal.
#Jika suhu melebihi 70.0°C, program menampilkan peringatan.
#Jika data tidak bisa diubah menjadi angka, ValueError ditangkap oleh except.
#Program menampilkan pesan data tidak valid, kemudian break menghentikan perulangan.
#Setelah perulangan selesai, program menampilkan jumlah data valid.
```

```python
#ALFREDO 4232511061 13/11/2026 PEMROGRAMAN DASAR
#SOAL BONUS

# Program Pemantauan Beban Listrik Harian PLTD
# Tujuan: Memantau kondisi beban listrik selama 5 hari

# Menyimpan data beban listrik dalam MW
beban = [75, 85, 95, 105, 120]

total_beban = 0

# Memeriksa beban listrik setiap hari
for i in range(len(beban)):
    try:
        nilai = float(beban[i])
        total_beban += nilai

        print(f"Hari ke-{i+1}: Beban = {nilai} MW")

        # Menentukan status beban listrik
        if nilai >= 110:
            print("STATUS: OVERLOAD")
        elif nilai >= 90:
            print("STATUS: WARNING")
        else:
            print("STATUS: NORMAL")

    except ValueError:
        print(f"Hari ke-{i+1}: Data tidak valid")

# Menghitung rata-rata beban
rata_rata = total_beban / len(beban)

print("-------------------------")
print(f"Total beban = {total_beban:.2f} MW")
print(f"Rata-rata beban = {rata_rata:.2f} MW")

#Program menyimpan data beban listrik selama 5 hari ke dalam list beban.
#Variabel total_beban digunakan untuk menampung jumlah seluruh beban.
#Perulangan for membaca data beban satu per satu.
#float() mengubah nilai menjadi angka desimal.
#Program menjumlahkan setiap beban ke dalam total_beban.
#Percabangan if-elif-else menentukan status beban:
#- Beban 110 MW atau lebih: OVERLOAD.
#- Beban 90 MW sampai kurang dari 110 MW: WARNING.
#- Beban di bawah 90 MW: NORMAL.
#try-except digunakan untuk menangani kesalahan konversi data.
#Setelah perulangan selesai, program menghitung dan menampilkan total serta rata-rata beban.
```
